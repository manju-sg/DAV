```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

Load both datasets

```python
ember = pd.read_csv(
    "../datasets/release_generation_yearly_global.csv"
)

owid = pd.read_csv(
    "../datasets/owid_energy.csv"
)

print("Ember:", ember.shape)
print("OWID:", owid.shape)
```

Extract India-specific data

```python
ember_india = ember[
    ember["Area"] == "India"
].copy()

owid_india = owid[
    owid["country"] == "India"
].copy()

print("Ember India:", ember_india.shape)
print("OWID India:", owid_india.shape)
```

Inspect data quality

```python
# Missing values
print("EMBER MISSING VALUES")
print(ember_india.isnull().sum())

print("\nOWID MISSING VALUES")
print(owid_india.isnull().sum())
```

Check duplicates:

```python
print("Ember duplicates:",
      ember_india.duplicated().sum())

print("OWID duplicates:",
      owid_india.duplicated().sum())
```

Check data types:

```python
ember_india.info()
owid_india.info()
```

Check statistical summaries:

```python
display(ember_india.describe())
display(owid_india.describe())
```

Select relevant columns

```python
owid_columns = [
    "country",
    "year",
    "population",
    "gdp",
    "electricity_demand_twh",
    "electricity_generation_twh",
    "electricity_emissions_mtco2eq",
    "electricity_carbon_intensity_gco2eq_per_kwh",
    "coal_electricity_twh",
    "gas_electricity_twh",
    "oil_electricity_twh",
    "solar_electricity_twh",
    "wind_electricity_twh",
    "hydro_electricity_twh",
    "nuclear_electricity_twh",
    "renewables_electricity_twh",
    "fossil_fuels_electricity_twh",
    "renewables_electricity_share_pct"
]

owid_clean = owid_india[owid_columns].copy()

display(owid_clean.head())
```

```python
ember_columns = [
    "Area",
    "Year",
    "Electricity source",
    "Is aggregated source",
    "Generation (TWh)",
    "Share of generation (%)",
    "Capacity (GW)",
    "Emissions (MtCO2e)",
    "Emissions intensity (gCO2e/kWh)"
]

ember_clean = ember_india[ember_columns].copy()

display(ember_clean.head())
```

Handle duplicate records

```python
print(
    ember_clean.duplicated(
        subset=["Area", "Year", "Electricity source"]
    ).sum()
)

print(
    owid_clean.duplicated(
        subset=["country", "year"]
    ).sum()
)
```

Sort and save cleaned datasets

```python
owid_clean = owid_clean.sort_values("year")

ember_clean = ember_clean.sort_values(
    ["Year", "Electricity source"]
)

owid_clean.to_csv(
    "../datasets/owid_india_clean.csv",
    index=False
)

ember_clean.to_csv(
    "../datasets/ember_india_clean.csv",
    index=False
)

print("Cleaned datasets saved!")
```
