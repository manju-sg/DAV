Load and prepare the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../datasets/owid_india_clean.csv")

df = df.sort_values("year").reset_index(drop=True)

df.head()
```

Calculate renewable energy's contribution to additional generation.

This compares how much renewable generation and coal generation increased over time.

```python
df["renewable_addition"] = df["renewables_electricity_twh"].diff()
df["coal_addition"] = df["coal_electricity_twh"].diff()
df["total_addition"] = df["electricity_generation_twh"].diff()

plt.figure(figsize=(12, 6))

plt.plot(df["year"], df["renewable_addition"],
         label="Renewable addition")

plt.plot(df["year"], df["coal_addition"],
         label="Coal addition")

plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("Year")
plt.ylabel("Additional generation (TWh)")
plt.title("Yearly Increase in Renewable and Coal Generation")
plt.legend()
plt.grid()
plt.show()
```

What this tells us: Whether renewable energy is contributing increasingly to India's additional electricity generation, or whether coal is still expanding faster.

Calculate renewable contribution to demand growth.

```python
df["demand_growth"] = df["electricity_demand_twh"].diff()
df["renewable_contribution"] = (
    df["renewable_addition"] / df["demand_growth"]
) * 100

plt.figure(figsize=(12, 6))

plt.plot(
    df["year"],
    df["renewable_contribution"],
    marker="o"
)

plt.axhline(100, linestyle="--", color="red")
plt.xlabel("Year")
plt.ylabel("Renewable contribution (%)")
plt.title("Renewable Generation Increase vs Electricity Demand Increase")
plt.grid()
plt.show()
```

Interpret carefully: this is a comparison of changes, not a direct measurement of how much demand was physically supplied by renewables. It can exceed 100% or become negative.

Analyze the relationship between renewable share and carbon intensity

```python
df["renewable_share"] = (
    df["renewables_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

plt.figure(figsize=(9, 6))

sns.regplot(
    data=df,
    x="renewable_share",
    y="electricity_carbon_intensity_gco2eq_per_kwh"
)

plt.xlabel("Renewable Share (%)")
plt.ylabel("Carbon Intensity (gCO2e/kWh)")
plt.title("Renewable Share vs Electricity Carbon Intensity")
plt.grid()
plt.show()

print(df[[
    "renewable_share",
    "electricity_carbon_intensity_gco2eq_per_kwh"
]].corr())
```

What this tells us: Whether years with a higher renewable share tend to have lower electricity carbon intensity. This is an association, not proof of causation.

Calculate compound annual growth rate (CAGR)

```python
def calculate_cagr(start_value, end_value, years):
    if start_value <= 0 or end_value < 0 or years <= 0:
        return np.nan

    return ((end_value / start_value) ** (1 / years) - 1) * 100

start_year = df["year"].min()
end_year = df["year"].max()
duration = end_year - start_year

for col in [
    "renewables_electricity_twh",
    "solar_electricity_twh",
    "wind_electricity_twh",
    "coal_electricity_twh",
    "electricity_demand_twh"
]:

    valid = df[["year", col]].dropna()

    start_value = valid.iloc[0][col]
    end_value = valid.iloc[-1][col]

    years = valid.iloc[-1]["year"] - valid.iloc[0]["year"]

    if start_value > 0 and end_value > 0 and years > 0:
        cagr = ((end_value / start_value) ** (1 / years) - 1) * 100
        print(f"{col}: {cagr:.2f}%")
    else:
        print(f"{col}: CAGR cannot be calculated")
```
