Import libraries and load cleaned data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

owid = pd.read_csv("../datasets/owid_india_clean.csv")
ember = pd.read_csv("../datasets/ember_india_clean.csv")

print(owid.shape)
print(ember.shape)
```

Analyse electricity generation trends.

Research question: How has India's electricity generation changed over time?

```python
plt.figure(figsize=(12, 6))

plt.plot(
    owid["year"],
    owid["electricity_generation_twh"],
    linewidth=2
)

plt.title("India's Electricity Generation Over Time")
plt.xlabel("Year")
plt.ylabel("Electricity Generation (TWh)")
plt.show()
```

Renewable energy growth.

Research question: Which renewable energy sources are growing?

```python
plt.figure(figsize=(12, 6))

plt.plot(owid["year"], owid["solar_electricity_twh"], label="Solar")
plt.plot(owid["year"], owid["wind_electricity_twh"], label="Wind")
plt.plot(owid["year"], owid["hydro_electricity_twh"], label="Hydro")

plt.title("Renewable Electricity Growth in India")
plt.xlabel("Year")
plt.ylabel("Generation (TWh)")
plt.legend()
plt.show()
```

Coal vs renewable generation.

Research question: How dependent is India on coal compared with renewable energy?

```python
plt.figure(figsize=(12, 6))

plt.plot(owid["year"], owid["coal_electricity_twh"], label="Coal")
plt.plot(owid["year"], owid["renewables_electricity_twh"], label="Renewables")

plt.title("Coal vs Renewable Electricity Generation")
plt.xlabel("Year")
plt.ylabel("Generation (TWh)")
plt.legend()
plt.show()
```

Electricity emissions.

Research question: How have electricity-related emissions changed?

```python
plt.figure(figsize=(12, 6))

plt.plot(
    owid["year"],
    owid["electricity_emissions_mtco2eq"]
)

plt.title("India's Electricity-Related Carbon Emissions")
plt.xlabel("Year")
plt.ylabel("Emissions (MtCO2e)")
plt.show()
```

Electricity carbon intensity.

Research question: Is electricity becoming cleaner per unit generated?

```python
plt.figure(figsize=(12, 6))

plt.plot(
    owid["year"],
    owid["electricity_carbon_intensity_gco2eq_per_kwh"]
)

plt.title("Electricity Carbon Intensity in India")
plt.xlabel("Year")
plt.ylabel("gCO2e/kWh")
plt.show()
```
