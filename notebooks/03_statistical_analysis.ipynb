Load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../datasets/owid_india_clean.csv")

df = df.sort_values("year").reset_index(drop=True)

print(df.shape)
display(df.head())
```

Calculate renewable energy growth rate.

Research question: How quickly is renewable electricity generation increasing?

```python
df["renewable_growth"] = (
    df["renewables_electricity_twh"]
    .pct_change() * 100
)

plt.figure(figsize=(12, 6))

plt.bar(df["year"], df["renewable_growth"])

plt.title("Yearly Renewable Energy Growth Rate in India")
plt.xlabel("Year")
plt.ylabel("Growth Rate (%)")
plt.show()
```

Calculate renewable electricity share.

Research question: What proportion of India's electricity generation comes from renewables?

```python
df["renewable_share"] = (
    df["renewables_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

plt.figure(figsize=(12, 6))

plt.plot(df["year"], df["renewable_share"])

plt.title("Renewable Electricity Share in India")
plt.xlabel("Year")
plt.ylabel("Renewable Share (%)")
plt.show()
```

Calculate coal dependency

```python
df["coal_share"] = (
    df["coal_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

plt.figure(figsize=(12, 6))

plt.plot(df["year"], df["coal_share"])

plt.title("India's Coal Dependency Over Time")
plt.xlabel("Year")
plt.ylabel("Coal Share (%)")
plt.show()
```

Now compare renewable share and coal share:

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["year"],
    df["renewable_share"],
    label="Renewables"
)

plt.plot(
    df["year"],
    df["coal_share"],
    label="Coal"
)

plt.title("Renewable vs Coal Share in India")
plt.xlabel("Year")
plt.ylabel("Share (%)")
plt.legend()
plt.show()
```

Analyse carbon intensity.

Research question: Is India producing electricity with fewer emissions per unit?

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["year"],
    df["electricity_carbon_intensity_gco2eq_per_kwh"]
)

plt.title("India's Electricity Carbon Intensity")
plt.xlabel("Year")
plt.ylabel("gCO2e/kWh")
plt.show()
```

Correlation analysis.

This is where we start investigating relationships between variables.

```python
columns = [
    "gdp",
    "population",
    "electricity_demand_twh",
    "electricity_generation_twh",
    "coal_electricity_twh",
    "solar_electricity_twh",
    "wind_electricity_twh",
    "hydro_electricity_twh",
    "electricity_emissions_mtco2eq",
    "electricity_carbon_intensity_gco2eq_per_kwh"
]

corr = df[columns].corr(method="pearson")

plt.figure(figsize=(14, 10))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between India's Energy and Economic Indicators")

plt.show()
```

GDP vs electricity demand.

Research question: How does India's economic development relate to electricity demand?

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="gdp",
    y="electricity_demand_twh",
    scatter_kws={"alpha": 0.6}
)

plt.title("GDP vs Electricity Demand in India")
plt.xlabel("GDP")
plt.ylabel("Electricity Demand (TWh)")

plt.show()
```
