Load the dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../datasets/owid_india_clean.csv")

df = df.sort_values("year").reset_index(drop=True)

print(df.shape)
display(df.head())
```

Create dashboard metrics

```python
df["renewable_share"] = (
    df["renewables_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

df["coal_share"] = (
    df["coal_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

df["solar_share"] = (
    df["solar_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

df["wind_share"] = (
    df["wind_electricity_twh"] /
    df["electricity_generation_twh"]
) * 100

df["renewable_growth"] = (
    df["renewables_electricity_twh"].pct_change() * 100
)

display(df[[
    "year",
    "renewable_share",
    "coal_share",
    "solar_share",
    "wind_share",
    "renewable_growth"
]].tail(10))
```

Create a summary table

```python
summary = df[[
    "electricity_generation_twh",
    "electricity_demand_twh",
    "renewables_electricity_twh",
    "coal_electricity_twh",
    "electricity_emissions_mtco2eq",
    "electricity_carbon_intensity_gco2eq_per_kwh"
]].describe()

display(summary)
```

Export dashboard data

```python
df.to_csv("../datasets/dashboard_data.csv", index=False)

print("Dashboard dataset exported successfully!")
```
