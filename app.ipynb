```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
ember = pd.read_csv(
    r"C:\Users\Manjunath\OneDrive\Documents\Python_Learning\DAV\datasets\release_generation_yearly_global.csv"
)

owid = pd.read_csv(
    r"C:\Users\Manjunath\OneDrive\Documents\Python_Learning\DAV\datasets\owid_energy.csv"
)

# Extract India
ember_india = ember[ember["Area"] == "India"].copy()
owid_india = owid[owid["country"] == "India"].copy()

# Sort by year
ember_india = ember_india.sort_values("Year")
owid_india = owid_india.sort_values("year")

# Display dimensions
print("Ember India:", ember_india.shape)
print("OWID India:", owid_india.shape)

# Display first five records
display(ember_india.head())
display(owid_india.head())
```

```python
energy = owid_india[
    ["year", "solar_electricity_twh",
     "wind_electricity_twh",
     "hydro_electricity_twh"]
].dropna()

energy.set_index("year").plot(figsize=(12, 6))

plt.title("India's Renewable Electricity Growth")
plt.xlabel("Year")
plt.ylabel("Electricity Generation (TWh)")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    owid_india["year"],
    owid_india["coal_electricity_twh"],
    label="Coal"
)

plt.plot(
    owid_india["year"],
    owid_india["renewables_electricity_twh"],
    label="Renewables"
)

plt.title("Coal vs Renewable Electricity Generation in India")
plt.xlabel("Year")
plt.ylabel("Generation (TWh)")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    owid_india["year"],
    owid_india["electricity_emissions_mtco2eq"]
)

plt.title("India's Electricity-Related Carbon Emissions")
plt.xlabel("Year")
plt.ylabel("Emissions (MtCO2e)")
plt.grid()
plt.show()
```
